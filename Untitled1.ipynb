```python
import pandas as pd
from io import StringIO

# Create sample table data
data = """Name Age City Marks
Amit 20 Delhi 85
Neha 21 Mumbai 90
Rahul 22 Pune 78
Priya 19 Agra 88"""

# Read the table into a DataFrame
copied_df = pd.read_csv(StringIO(data), sep=r'\s+')

# Display column names
print(copied_df.columns)

# Create a new DataFrame with three columns
new_df = copied_df[['Name', 'Age', 'City']]

print(new_df)
```

```python
import pandas as pd
import numpy as np

data = {
    'Student': ['Amit', 'Neha', 'Rahul', 'Priya', 'Karan'],
    'Course': ['Python', 'Java', 'C++', 'Python', 'Java'],
    'Score': [85, 90, 78, 88, 92]
}

course_df = pd.DataFrame(data)

# Display column names
print(course_df.columns)

# Retrieve Student column
print(course_df['Student'])

# Create a DataFrame with Student and Score
new_df = course_df[['Student', 'Score']]
print(new_df)

# Add Batch column
course_df['Batch'] = '2026'
print(course_df)
```

```python
data = {
    'Destination': ['Delhi', 'Mumbai', 'Jaipur', 'Pune', 'Agra'],
    'Days': [2, 4, 3, 5, 1]
}

travel_df = pd.DataFrame(data)

# Add Code using np.arange(5)
travel_df['Code'] = np.arange(5)
print(travel_df)

# Replace Code with a Series at indexes 1 and 4
codes = pd.Series([101, 104], index=[1, 4])
travel_df['Code'] = codes

print(travel_df)
```

```python
data = {
    'Student': ['Aman', 'Riya', 'Kunal', 'Simran', 'Rohit'],
    'Class': [10, 10, 9, 9, 10],
    'Section': ['A', 'B', 'A', 'B', 'A']
}

school_df = pd.DataFrame(data)

# Request a column that does not exist
school_df = pd.DataFrame(
    school_df,
    columns=['Student', 'Class', 'Section', 'House']
)

print(school_df)

# Assign one common value to House
school_df['House'] = 'Red House'

print(school_df)
```

```python
data = {
    'Item': ['Pen', 'Book', 'Pencil', 'Bag', 'Eraser'],
    'Quantity': [20, 15, 30, 5, 25],
    'Price': [10, 50, 5, 500, 3]
}

inventory_df = pd.DataFrame(data)

# Add a partially indexed Series
inventory_df['Rack'] = pd.Series(
    ['A1', 'B2', 'C3'],
    index=[0, 2, 4]
)

# Display DataFrame
print(inventory_df)

# Delete the added column
del inventory_df['Rack']

# Display DataFrame again
print(inventory_df)
```

```python
import pandas as pd

# Create a DataFrame instead of reading from clipboard
data = {
    'Name': ['Amit', 'Neha', 'Rahul', 'Priya'],
    'City': ['Delhi', 'Mumbai', 'Pune', 'Agra'],
    'Total Marks': [85, 90, 78, 88],
    'Age': [20, 21, 19, 22]
}

lab_df = pd.DataFrame(data)

# Display the complete DataFrame
print(lab_df)

# Display column names
print(lab_df.columns)

# Retrieve a single-word column
print(lab_df['Name'])

# Retrieve a column containing a space
print(lab_df['Total Marks'])

# Create a smaller DataFrame
small_df = lab_df[['Name', 'City', 'Total Marks']]
print(small_df)

# Add a new column
lab_df['Batch'] = '2026'
print(lab_df)

# Delete the new column
del lab_df['Batch']
print(lab_df)
```
