```python

import pandas as pd

# Create a labelled Series
sales = pd.Series(
    [450, 650, 300, 800],
    index=['Pen', 'Bag', 'Book', 'Laptop']
)

# Display values
print("Values:")
print(sales.values)

# Display index
print("\nIndex:")
print(sales.index)

# Select one label
print("\nSales of Bag:")
print(sales['Bag'])

# Filter sales above 500
print("\nSales above 500:")
print(sales[sales > 500])

# Check whether a label exists
print("\nIs Pen present?")
print('Pen' in sales.index)

# Convert Series to dictionary
sales_dict = sales.to_dict()
print("\nSales Dictionary:")
print(sales_dict)
```

```python

import pandas as pd

data = {'A': 15, 'B': 35, 'C': 25}

s = pd.Series(data, index=['C', 'A', 'D', 'B'])

s.name = 'Observations'
s.index.name = 'Category'

print(s)
print("\nBoolean filter:")
print(s[s > 20])

print("\nNon-missing values:")
print(pd.notnull(s))
```
